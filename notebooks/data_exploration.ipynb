```python
import pandas as pd
```

```python
df = pd.read_parquet("../data/raw/green_tripdata_2025-03.parquet")
df.head()
```

```python
df.columns
```

```python
selected_columns = ["lpep_pickup_datetime", "total_amount"]

selected_df = df.copy()

selected_df = selected_df[selected_columns]

print(selected_df.shape)
print(selected_df.info())
print(selected_df.describe())
print(selected_df.head())
```

```python
# pick only march rides
print(selected_df.shape)
p = selected_df["lpep_pickup_datetime"]
selected_df = selected_df[(p >= "2025-03-01") & (p < "2025-04-01")]
print(selected_df.shape)
```

```python
daily_revenue = (
    selected_df.groupby(selected_df["lpep_pickup_datetime"].dt.date)["total_amount"].sum()
)
daily_revenue.head()
```
